```python
#import relevant frameworks
import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error
```

```python
#load dataset
df=pd.read_csv("Salary_Data.csv")
df.head()
```

```python
df.info()
```

```python
#determine features and variables
x=df[["YearsExperience"]]
y=df["Salary"]
```

#splitting the dataset
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.20,random_state=42)

#justification
An 80:20 split was choosen to provide enough training data while reserving 20% for testing the model.

```python
#train the model
from sklearn.linear_model import LinearRegression
model=LinearRegression()
model.fit(x_train,y_train)
```

```python
#prediction
y_predict=model.predict(x_test)
print(y_predict)
```

```python
#Evaluate the model
mse=mean_squared_error(y_test,y_predict)
print(mse)
```

Since lower mse values indicate better performance, this indicates that the model's predivtion is reasonably good.

```python
#fitted regression equation
intercept=model.intercept_
slope=model.coef_[0]
print("Intercept:",intercept)
print("Slope:",slope)
```

slope: for every additional year of experience,the salary is predicted to increase by 9,423.82
intercept: when the candidate has no years of experience, the model predicts a salary of 25,321.6

```python
#visualize the Regression Line
import matplotlib.pyplot as plt
plt.scatter(x,y,color="blue",label="Actual Data")
plt.plot(x,model.predict(x),color="red",linewidth=2,label="Regression Line")
plt.xlabel("Years of Experience")
plt.ylabel("Salary")
plt.title("Actual Data and Fitted Regression Line")
plt.legend()
plt.show()
```
